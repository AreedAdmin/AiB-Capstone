```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
demand_df = pd.read_csv('Residential_demand.csv')
turbine_df = pd.read_csv('Turbine_telemetry.csv')
```

```python
turbine_df.Timestamp = pd.to_datetime(turbine_df.Timestamp)
turbine_df.head()
```

```python
demand_df.Timestamp = pd.to_datetime(demand_df.Timestamp)
demand_df.head()
```

```python
plt.plot(demand_df['Timestamp'], demand_df['N_households'])
plt.xlabel('Timestamp')
plt.ylabel('Number of Households')
plt.title('Number of Households Over Time')
plt.show()
```

```python
plt.plot(demand_df['Timestamp'], demand_df['Demand_mean_kw'])
plt.xlabel('Timestamp')
plt.ylabel('Mean Demand (kW)')
plt.title('Residential Demand Over Time')
plt.show()
```

```python
plt.plot(demand_df['Timestamp'], demand_df['Demand_mean_kw']*demand_df['N_households'])
plt.xlabel('Timestamp')
plt.ylabel('Total Demand (kW)')
plt.title('Total Demand Over Time')
plt.show()
```

```python
plt.plot(turbine_df['Timestamp'], turbine_df['Setpoint_kw'])
plt.xlabel('Timestamp')
plt.ylabel('Setpoint (kW)')
plt.title('Turbine Setpoints Over Time')
plt.show()
```

```python
plt.plot(turbine_df['Timestamp'], turbine_df['Power_kw'])
plt.xlabel('Timestamp')
plt.ylabel('Power (kW)')
plt.title('Turbine Power Output Over Time')
plt.show()
```

```python
plt.plot(turbine_df['Timestamp'], turbine_df['Wind_ms'])
plt.xlabel('Timestamp')
plt.ylabel('Wind Speed (m/s)')
plt.title('Wind Speed Over Time')
plt.show()
```
